```python
from PIL import Image

def create_colored_wallpaper(rgb_color, output_filename="plain_wallpaper.png"):
    width, height = 1080, 2340  # Pixel 4a 5G resolution
    image = Image.new("RGB", (width, height), rgb_color)
    image.save(output_filename)
    print(f"Plain wallpaper saved as {output_filename}")

# Example usage
if __name__ == "__main__":
    create_colored_wallpaper(
        rgb_color=(223, 177, 213)
    )
```
